### Learning goals
Store values, identify common types, calculate a mean, and convert numeric text.

A variable is a name referring to a value. `=` assigns a value. Python names are case-sensitive. Use descriptive names such as `passing_grade`.


````python
name = "Maya"      # str: text
grade = 88
weight = 0.5       # float: decimal
is_present = True  # bool: True or False
print(name, grade, weight, is_present)
print(type(name).__name__, type(grade).__name__)
print(type(weight).__name__, type(is_present).__name__)
````

### Arithmetic
`+`, `-`, `*`, `/` perform addition, subtraction, multiplication, and division. `**` raises a power, `//` gives floor division, and `%` gives the remainder. Parentheses control grouping.

## 04 — Decisions and Repetition

**Python Foundations · Slides 19–30 · Python 3.10+**

[Back to contents](#contents)

### Learning goals
Choose behavior with `if`, repeat work with `for`, and calculate a summary.

Use a colon to begin a block and four spaces for its body. `if` runs its body when the condition is true. `else` handles the remaining case. The passing threshold of 75 is only a teaching example.


````python
grade = 75
if grade >= 75:
    print("Pass")
else:
    print("Needs practice")
````

**Expected:** `Pass`. Predict the result for 74, then restore 75.


### A loop visits each item
Each repetition assigns the next dictionary to `student`. Indentation determines what repeats.


````python
students = [
    {"name": "Maya", "grade": 88},
    {"name": "Luis", "grade": 72},
    {"name": "Ari", "grade": 95},
    {"name": "Bea", "grade": 75},
]
for student in students:
    print(student["name"], student["grade"])
print("Finished")
````

### Step-by-step trace
The next cell shows each iteration explicitly. `enumerate(..., start=1)` supplies a human-readable iteration number. Watch how `total` changes and where the loop ends.

Expected totals: 88, 160, 255, 330. `Finished` appears once, after all four iterations.


````python
grades = [88, 72, 95, 75]
total = 0
for step, grade in enumerate(grades, start=1):
    previous_total = total
    total += grade
    print(f"Step {step}: grade={grade}; total={previous_total}+{grade}={total}")
print("Finished")
````

**Expected:** `Maya 88`, `Luis 72`, `Ari 95`, `Bea 75`, then `Finished`, each on a new line.


### Accumulation
Initialize `total` before the loop. `+=` adds to the previous value. Place the average calculation after the loop.


````python
total = 0
passing_names = []
for student in students:
    total += student["grade"]
    if student["grade"] >= 75:
        passing_names.append(student["name"])
average = total / len(students)
print(f"Average: {average:.2f}")
print(passing_names)
````

**Expected:** `Average: 82.50` and `['Maya', 'Ari', 'Bea']`.

## 05 — Reusable Functions

**Python Foundations · Slides 31–38 · Python 3.10+**

[Back to contents](#contents)

### Learning goals
Define a function, pass arguments, return a result, and recognize common errors.

A function groups reusable instructions. A **parameter** is a name in its definition. An **argument** is a value supplied when calling it. Defining a function does not execute its body.


### Four function patterns
No parameters/no explicit return: `show_heading`. With a parameter/no explicit return: `greet_student`. No parameters/with return: `default_threshold`. With parameters/with return: `add_bonus`.

Expected output: `Student grade report`, `Hello, Maya!`, `75`, `75`, `Hello, Luis!`, and `None`, each on a new line.


````python
# Four patterns: inputs and return values are independent.
def show_heading():
    print("Student grade report")

def greet_student(name):
    print(f"Hello, {name}!")

def default_threshold():
    return 75

def add_bonus(grade, bonus):
    return grade + bonus

show_heading()
greet_student("Maya")
print(default_threshold())
print(add_bonus(72, 3))
result = greet_student("Luis")
print(result)  # No explicit return means the returned value is None.
````

````python
def mean_grade(grades):
    if not grades:
        return None
    return sum(grades) / len(grades)

average = mean_grade([88, 72, 95, 75])
print(f"Average: {average:.2f}")
print(mean_grade([]))
````

**Expected:** `Average: 82.50` and `None`. `None` represents no result here. The empty-list check prevents division by zero.

`return` sends a value to the caller. `print()` displays a value. A function that only prints returns `None` implicitly.


````python
def is_passing(grade, threshold=75):
    return grade >= threshold

print(is_passing(75))
print(is_passing(75, threshold=80))
````

**Expected:** `True` and `False`. `75` is an illustrative default threshold.


### Errors are feedback
`SyntaxError`: Python cannot parse the statement. `NameError`: a name has no value. `TypeError`: an operation uses incompatible types. `ValueError`: a value cannot be converted as requested.

The next cell catches a deliberate conversion error so Run all still succeeds. Read the final traceback line first when debugging your own code.


````python
try:
    grade = float("absent")
except ValueError:
    print("Please use a numeric grade.")
````

### Complete analyzer functions
This cell is the same implementation used by `examples/grade_analyzer.py`. First follow `ensure_sample_data`, `read_students`, `summarize`, and `write_report`. Validation inside `read_students` is reference material: it gives helpful feedback for invalid grades and CSV headers.

`csv.DictReader` gives one dictionary per row. CSV values arrive as strings. `float()` converts each grade. `newline=""` lets the CSV module handle line endings correctly.


````python
from pathlib import Path
import csv
import math

PASSING_GRADE = 75  # Illustrative teaching rule, not institutional policy.
SAMPLE_CSV = "name,grade\nMaya,88\nLuis,72\nAri,95\nBea,75\n"


def ensure_sample_data(path):
    """Create fictional sample data only when the file is missing."""
    path.parent.mkdir(parents=True, exist_ok=True)
    if not path.exists():
        with path.open("w", encoding="utf-8", newline="") as file:
            file.write(SAMPLE_CSV)


def read_students(path):
    """Read name/grade CSV records and reject incomplete or invalid data."""
    students = []
    with path.open("r", encoding="utf-8", newline="") as file:
        reader = csv.DictReader(file)
        if reader.fieldnames != ["name", "grade"]:
            raise ValueError("CSV headers must be: name,grade")
        for row_number, row in enumerate(reader, start=2):
            name = (row.get("name") or "").strip()
            try:
                grade = float(row["grade"])
            except (TypeError, ValueError) as error:
                raise ValueError(f"Row {row_number}: grade must be a number") from error
            if not name or None in row or not math.isfinite(grade) or not 0 <= grade <= 100:
                raise ValueError(f"Row {row_number}: use a name and a grade from 0 to 100")
            students.append({"name": name, "grade": grade})
    return students


def summarize(students, passing_grade=PASSING_GRADE):
    """Return a report, including a helpful result for an empty class."""
    if not students:
        return "No student records.\n"
    total = 0
    passing_names = []
    for student in students:
        total += student["grade"]
        if student["grade"] >= passing_grade:
            passing_names.append(student["name"])
    average = total / len(students)
    names = ", ".join(passing_names) if passing_names else "None"
    return (
        f"Students: {len(students)}\n"
        f"Class average: {average:.2f}\n"
        f"Passing threshold (illustrative): {passing_grade}\n"
        f"Passing students: {len(passing_names)}\n"
        f"Names: {names}\n"
    )


def write_report(path, report):
    """Replace the generated report on every run."""
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8") as file:
        file.write(report)
````

### Reading student records
The CSV must have `name,grade` headers and grades from 0 to 100. A header-only file represents an empty class. Missing files get fictional sample data.


````python
csv_path = data_dir / "grades.csv"
ensure_sample_data(csv_path)
students = read_students(csv_path)
for student in students:
    print(student["name"], student["grade"])
````

**Expected with bundled data:** `Maya 88.0`, `Luis 72.0`, `Ari 95.0`, `Bea 75.0` on separate lines.


### Analysis and writing
The same input must produce the same summary in the notebook and script. The report uses a trailing newline to keep text files readable.


````python
report = summarize(students)
report_path = output_dir / "grade_summary.txt"
write_report(report_path, report)
print(report, end="")
print("Saved:", report_path)
with report_path.open("r", encoding="utf-8") as file:
    saved_report = file.read()
assert saved_report == report
````

**Expected with bundled data**
```text
Students: 4
Class average: 82.50
Passing threshold (illustrative): 75
Passing students: 3
Names: Maya, Ari, Bea
```
The final `Saved:` line shows your runtime-specific path.
