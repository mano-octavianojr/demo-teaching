## Learning goals
Define a function, pass arguments, return a result, and recognize common errors.

A function groups reusable instructions. A **parameter** is a name in its definition. An **argument** is a value supplied when calling it. Defining a function does not execute its body.


## Four function patterns
No parameters/no explicit return: `show_heading`. With a parameter/no explicit return: `greet_student`. No parameters/with return: `default_threshold`. With parameters/with return: `add_bonus`.

Expected output: `Student grade report`, `Hello, Maya!`, `75`, `75`, `Hello, Luis!`, and `None`, each on a new line.


```python
# Four patterns: inputs and return values are independent.
def show_heading():
    print("Student grade report")

def greet_student(name):
    print(f"Hello, {name}!")

def default_threshold():
    return 75

def add_bonus(grade, bonus):
    return grade + bonus

show_heading()
greet_student("Maya")
print(default_threshold())
print(add_bonus(72, 3))
result = greet_student("Luis")
print(result)  # No explicit return means the returned value is None.
```

```python
def mean_grade(grades):
    if not grades:
        return None
    return sum(grades) / len(grades)

average = mean_grade([88, 72, 95, 75])
print(f"Average: {average:.2f}")
print(mean_grade([]))
```

**Expected:** `Average: 82.50` and `None`. `None` represents no result here. The empty-list check prevents division by zero.

`return` sends a value to the caller. `print()` displays a value. A function that only prints returns `None` implicitly.


```python
def is_passing(grade, threshold=75):
    return grade >= threshold

print(is_passing(75))
print(is_passing(75, threshold=80))
```

**Expected:** `True` and `False`. `75` is an illustrative default threshold.


## Errors are feedback
`SyntaxError`: Python cannot parse the statement. `NameError`: a name has no value. `TypeError`: an operation uses incompatible types. `ValueError`: a value cannot be converted as requested.

The next cell catches a deliberate conversion error so Run all still succeeds. Read the final traceback line first when debugging your own code.


```python
try:
    grade = float("absent")
except ValueError:
    print("Please use a numeric grade.")
```

**Expected:** `Please use a numeric grade.`

Write `grade_label(grade)` that returns `"Pass"` for grades at least 75 and `"Needs practice"` otherwise. Call it with 75 and 72.

## Solution


```python
def grade_label(grade):
    if grade >= 75:
        return "Pass"
    return "Needs practice"

print(grade_label(75))
print(grade_label(72))
```

**Expected:** `Pass` and `Needs practice`.

**Exit check:** What does a function return if it reaches the end without `return`? `None`.
