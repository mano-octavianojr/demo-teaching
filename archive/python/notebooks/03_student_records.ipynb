## Learning goals
Work with strings, lists, and dictionaries. Choose a collection that fits the information.

Strings hold text. A **method** is an operation called through a value, such as `name.strip()`. Strings are immutable: their methods return new strings.


```python
name = "  Maya  "
clean_name = name.strip()
print(clean_name)
print(clean_name.upper())
print(f"Student: {clean_name}")
```

**Expected:** `Maya`, `MAYA`, and `Student: Maya`.


## Lists: an ordered collection
List positions start at zero. `grades[0]` reads the first element. `append()` changes the list by adding one item.


```python
grades = [88, 72, 95]
print(grades[0])
grades.append(75)
print(grades)
print(len(grades))
print(grades[-1])
```

**Expected:** `88`, `[88, 72, 95, 75]`, `4`, and `75`. Index `-1` refers to the last element.


## Dictionaries: named fields
A dictionary maps keys to values. Use a list of dictionaries for several students, so each name stays beside its grade.


```python
student = {"name": "Maya", "grade": 88}
print(student["name"], student["grade"])
students = [student, {"name": "Luis", "grade": 72}]
print(students[1]["name"])
print(students[1]["grade"])
```

**Expected:** `Maya 88`, `Luis`, and `72`.
