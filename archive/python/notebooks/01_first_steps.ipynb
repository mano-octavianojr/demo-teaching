A program is a set of instructions. Python runs these instructions in order. `print()` displays a value. Quotation marks make text a **string**.

A comment starts with `#` outside a string. Python ignores the comment when executing code.


````python
print("# This is text, not a comment")
print(88)
print("Maya", 88)
````

**Expected output:** `# This is text, not a comment`, then `88`, then `Maya 88`, on separate lines.
