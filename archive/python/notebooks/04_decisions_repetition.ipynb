## A loop visits each item
Each repetition assigns the next dictionary to `student`. Indentation determines what repeats.


```python
students = [
    {"name": "Maya", "grade": 88},
    {"name": "Luis", "grade": 72},
    {"name": "Ari", "grade": 95},
    {"name": "Bea", "grade": 75},
]
for student in students:
    print(student["name"], student["grade"])
print("Finished")
```

## Step-by-step trace
The next cell shows each iteration explicitly. `enumerate(..., start=1)` supplies a human-readable iteration number. Watch how `total` changes and where the loop ends.

Expected totals: 88, 160, 255, 330. `Finished` appears once, after all four iterations.


```python
grades = [88, 72, 95, 75]
total = 0
for step, grade in enumerate(grades, start=1):
    previous_total = total
    total += grade
    print(f"Step {step}: grade={grade}; total={previous_total}+{grade}={total}")
print("Finished")
```

**Expected:** `Maya 88`, `Luis 72`, `Ari 95`, `Bea 75`, then `Finished`, each on a new line.


## Accumulation
Initialize `total` before the loop. `+=` adds to the previous value. Place the average calculation after the loop.


```python
total = 0
passing_names = []
for student in students:
    total += student["grade"]
    if student["grade"] >= 75:
        passing_names.append(student["name"])
average = total / len(students)
print(f"Average: {average:.2f}")
print(passing_names)
```

**Expected:** `Average: 82.50` and `['Maya', 'Ari', 'Bea']`.

## Solution


```python
high_grade_count = 0
for student in students:
    if student["grade"] >= 90:
        high_grade_count += 1
print(high_grade_count)
```
