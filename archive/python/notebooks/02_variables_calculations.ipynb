## Learning goals
Store values, identify common types, calculate a mean, and convert numeric text.

A variable is a name referring to a value. `=` assigns a value. Python names are case-sensitive. Use descriptive names such as `passing_grade`.


```python
name = "Maya"      # str: text
grade = 88
weight = 0.5       # float: decimal
is_present = True  # bool: True or False
print(name, grade, weight, is_present)
print(type(name).__name__, type(grade).__name__)
print(type(weight).__name__, type(is_present).__name__)
```

**Expected:** `Maya 88 0.5 True`, then `str int`, then `float bool`.

## Arithmetic
`+`, `-`, `*`, `/` perform addition, subtraction, multiplication, and division. `**` raises a power, `//` gives floor division, and `%` gives the remainder. Parentheses control grouping.


```python
total = 88 + 72 + 95 + 75
average = total / 4
print(total)
print(average)
print(7 // 2, 7 % 2, 2 ** 3)
print(2 + 3 * 4, (2 + 3) * 4)
```

**Expected:** `330`, `82.5`, `3 1 8`, and `14 20`.

## Comparisons and conversion
`==` compares equality. `>=` means greater than or equal to. A comparison produces a Boolean. Text read from a CSV file needs conversion before arithmetic. `input()` also returns text, but these examples avoid interactive prompts so Run all works.


```python
grade_text = "88"
grade = float(grade_text)
print(grade + 2)
print(grade >= 75)
print(grade == 75)
print(f"Maya's grade: {grade:.1f}")
```

**Expected:** `90.0`, `True`, `False`, and `Maya's grade: 88.0`.
An f-string inserts values inside `{}`. `:.1f` displays one decimal place.
